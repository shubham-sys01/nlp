```
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk import RegexpParser
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")

text = "The boys and girls are playing the school ground along with the teachers"
grammar = r"""
NP:{<DT>?<JJ>*<NN.*>+}
PP:{<IN><NP>}
VP:{<VB.*><NP|PP>*}
"""
parser = RegexpParser(grammar)
tree = parser.parse(pos_tag(word_tokenize(text)))
print(tree)
tree.draw()
```

```
#lab 4 tfidf
import string
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

corpus = [
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?"
]

def preprocess(text):
  tokens = word_tokenize(text)
  tokens = [word for word in tokens if word.lower() not in string.punctuation]
  sw = set(stopwords.words("english"))
  tokens = [word for word in tokens if word.lower() not in sw]
  return " ".join(tokens)

documents = [preprocess(doc) for doc in corpus]
tfid = TfidfVectorizer()
matrix = tfid.fit_transform(documents)
print(tfid.get_feature_names_out())
print(matrix.toarray())
```

```
#lab 4b cyk
import nltk
grammar = nltk.CFG.fromstring("""
S ->NP VP
NP -> DET N
VP -> V NP
DET -> 'the'
N -> 'dog'|'cat'
V -> 'chased'
""")
parser = nltk.ChartParser(grammar)
text = "the dog chased the cat"
trees = list(parser.parse(text.split()))
if(trees):
  print("valid")
  for tree in trees:
    print(tree)
else:
  print("invalid")
```

```
#lab5a grams
import nltk
from nltk.util import ngrams
nltk.download("punkt")
nltk.download("punkt_tab")

text = "this is a sample text to test grams"
tokens = nltk.word_tokenize(text)
for n in range(1,4):
  print(f"\n{n}-ngrams")
  for grams in ngrams(tokens,n):
    print(grams)
```

```
from collections import Counter

corpus = [
    "This is a dog",
    "This is a cat",
    "I love my cat",
    "This is my name"
]

def getbigram(words):
  return [(words[i],words[i+1]) for i in range(len(words)-1)]

sentences = [sentence.lower().split() for sentence in corpus]
words = [word for sentence in sentences for word in sentence]
bigrams = [bg for sentence in sentences for bg in getbigram(sentence)]

unigramcount = Counter(words)
bigramcount = Counter(bigrams)

bigramprob = {
    bg : count / unigramcount[bg[0]]
    for bg , count in bigramcount.items()
}

prob = 1.0
test = "This is my cat"
testbigarm = getbigram(test.lower().split())

for bg in testbigarm:
  prob*= bigramprob.get(bg,0.0)

print("all bigrams", bigrams)
print("bigram prob" , bigramprob)
print("unigram count" , dict(unigramcount))
print("bigram count" , dict(bigramcount))
print(prob)
```

```
#lab6 stanford
import stanza
stanza.download('en',verbose=False)

nlp = stanza.Pipeline(lang='en',processors='tokenize,ner',verbose=False)
def namedenitity(text):
  doc = nlp(text)
  return [(ent.text,ent.type) for ent in doc.entities]

documents = [
    "Barack Obama was the 44th President of the United States.",
    "Microsoft was founded by Bill Gates in 1975.",
    "The Taj Mahal is located in India."
]

for i , doc in enumerate(documents):
  print(f"{i+1} document")
  print("named enitities: ",namedenitity(doc))
```

```
from collections import Counter
from nltk.corpus import stopwords , reuters
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer

nltk.download('stopwords')
nltk.download('reuters')
nltk.download('punkt')
nltk.download('punkt_tab')
doc = [reuters.raw(fid) for fid in reuters.fileids()[:100]]
def countstopwords():
  count = Counter()
  sw = set(stopwords.words('english'))

  count.update(word for sentence in doc for word in nltk.word_tokenize(sentence.lower()) if word in sw)
  return count

print("most common")
print(countstopwords().most_common(10))

tfid = TfidfVectorizer(stop_words='english')
matrix = tfid.fit_transform(doc)
words = tfid.get_feature_names_out()
scores = matrix[0].toarray()[0]
print(sorted(zip(words,scores),key=lambda item :item[1],reverse=True)[:10])
```

```
#lab 8 sentiment
import nltk
from nltk.corpus import movie_reviews
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

nltk.download("movie_reviews")

doc=[]
label =[]
categories = ['pos',"neg"]
for c in categories:
  for fileid in movie_reviews.fileids(c):
    doc.append(movie_reviews.raw(fileid))
    label.append(c)


model = make_pipeline(TfidfVectorizer(stop_words="english"),MultinomialNB())
model.fit(doc,label)
test_sentences = [
    "I absolutely loved this movie!",
    "This was the worst film I've ever seen.",
    "The storyline was decent, but the acting could have been better."
]

for sentence in test_sentences:
    prediction = model.predict([sentence])[0]
    print(sentence,prediction)

```
